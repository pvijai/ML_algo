# Linear Regression with OLS



```python
import numpy as np
import pandas as pd
import numpy.matlib
import matplotlib.pyplot as plt

# read dataset
from google.colab import files
uploaded=files.upload()

df = pd.read_csv("housing_price_data.csv")

df.head()
```

```python
# Plot data points
X, y = (df["Size"].values,df["Price"].values)
X = X.reshape(len(X),1)
y = y.reshape(len(y),1)

# plots the data points
plt.plot(X, y, 'ro')
plt.xlabel("# Square footage")
plt.ylabel("Price")
plt.show()
```

```python
# Normalize the house size, e.g., subtract min and divide by (max-min) to ensure
# the scale is compatible with the bias, which is 1
max_min = X.max(0) - X.min(0)
Xmin = X.min(0)
X = (X-Xmin) / max_min

# Add bias
z=np.ones((len(X),1))
X = X.reshape(len(X),1)
X = np.concatenate((z,X), axis=1)
y = y.reshape(len(X),1)

# sort for testing/plotting
Xsort = np.sort(X, axis=0)

print(X.shape)
print(y.shape)
```

## **OLS METHOD**

```python
# Linear Regression by Ordinary Least Squares method
# TO DO: compute w_ols

## Your code goes here ...

XTX=X.T @ X
x_inverse=np.linalg.inv(XTX)
w_ols=x_inverse @ X.T @ y
print(f"w_OLS = [{w_ols[0, 0]:.6f}, {w_ols[1, 0]:.6f}]")
```

```python
# TO DO: Predict the price of a 5000 sqft house
### Remember to normalize it first

## Your code goes here ...
house=5000
normalized_house=(house-Xmin)/(max_min)
x_house=np.array([[1,normalized_house[0]]])
predicted_price=x_house @ w_ols
print(f"Predicted price for 5000 sqft house is ${predicted_price[0,0]:,.2f}")
```

```python
### Plot fitted curve

yhat_ols = np.dot(np.sort(X, axis=0),w_ols)

plt.plot(X[:,1] * max_min + Xmin, y, 'ro')
plt.plot(Xsort[:,1] *max_min + Xmin, yhat_ols, 'g', label="Ordinary Least Squares")
plt.legend()
plt.xlabel("# Square footage")
plt.ylabel("Price")
plt.show()
```
