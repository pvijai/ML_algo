# DISCRIMINANT ANALYSIS

In this coding assignment you are to implement a Minimum Risk Bayes Decision Theoretic classifier. Use the training set to train the classifier and the validation set to evaluate the accuracy.

Assume the following:
1. All conditional density functions are multivariate Gaussian
2. Each class has its own covariance matrix
3. Equal prior probabilities
4. 0-1 loss function


## Load datasets

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Load training data - 135 observations, 4 features, 3 classes,
df = pd.read_csv("iris_corrupted_training_dataset.csv")
print(df.head())
df = df.values
train_data = df

# Load validation data - 15 samples
df = pd.read_csv("iris_validation_dataset.csv")
print(df.head())
df = df.values
val_data = df
```

```python
## Your code goes here ...

X=train_data[:,0:4]
Y=train_data[:,4]
X_val=val_data[:,0:4]
Y_val=val_data[:,4]
mean={}
covariance={}
classes=np.unique(Y)
prior_probability=1/(len(classes))

X_train={}

for i in classes:
  X_train[i]=X[Y==i]

#Calculate mean and covariance
for i in classes:
  mean[i]=np.mean(X_train[i],axis=0)
  covariance[i]=np.cov(X_train[i],rowvar=False)

  print(f"class {i}")
  print(f"mean: {mean[i]}")
  print(f"covariance: {covariance[i]}")


#Calculate determinant
def predict(x_value):
  log_pdf=np.zeros((x_value.shape[0],len(classes)))
  for x,val in enumerate(x_value):
    for i,c in enumerate(classes):
      d=X.shape[1]
      diff=val-mean[c]
      sign,logdet=np.linalg.slogdet(covariance[c])
      cov_diff=np.linalg.solve(covariance[c],diff)
      mahalanobis=np.dot(diff,cov_diff)
      log_pdf[x,i]= -0.5*(d*np.log(2*np.pi)) -0.5*(logdet) -.5*mahalanobis +np.log(prior_probability)
  return log_pdf

```

```python
#Validation
log_value=predict(X_val)
predicted_indices=np.argmax(log_value,axis=1)
predictedClass=classes[predicted_indices]

print("Results")
for i in range(len(predictedClass)):
  print(f"y={Y_val[i]}, y_hat={predictedClass[i]}")
  accuracy=(np.sum(Y_val==predictedClass))/len(Y_val)

print (accuracy)
print(f"Average classification accuracy: {accuracy}")
print(f"Average accuracy: {(accuracy*100):.4f}%")
```
